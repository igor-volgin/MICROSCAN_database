```python
import os

import numpy as np
import matplotlib.pyplot as plt
```

```python
folder = "DSC_data"

groups = {
    "PE": ["PE1.txt", "PE2.txt"],
    "PP": ["PP1.txt", "PP2.txt"],
    "PS": ["PS1.txt", "PS2.txt"],
}

data = {poly: [] for poly in groups}

for poly, files in groups.items():
    for fname in files:
        path = os.path.join(folder, fname)

        temps = []
        dsc = []

        with open(path, "r", encoding="latin1", errors="ignore") as f:
            for line in f:
                if line.startswith("#"):
                    continue

                parts = line.strip().split(";")
                if len(parts) < 5:
                    continue

                try:
                    T = float(parts[0])
                    dsc_val = float(parts[2])
                    segment = int(parts[4])
                except:
                    continue

                if segment == 5:
                    temps.append(T)
                    dsc.append(dsc_val)

        data[poly].append((np.array(temps), np.array(dsc)))

# ---- Plotting ----

fig, axes = plt.subplots(3, 1, figsize=(7, 5), sharex=True)

# y-axis minimum limits
ymin = {"PE": 0.0, "PP": 0.5, "PS": 2.2}

# bold panel letter labels
panel_labels = {"PE": "(a)", "PP": "(b)", "PS": "(c)"}

for ax, poly in zip(axes, ["PE", "PP", "PS"]):
    for i, (T, dsc) in enumerate(data[poly]):
        ax.plot(T, dsc, label=f"{poly}{i+1}")

    # bold (a), (b), (c)
    ax.text(0.02, 0.95, panel_labels[poly], transform=ax.transAxes,
            va="top", ha="left", fontsize=12, fontweight="bold")

    ax.legend(loc="upper right")
    ax.grid(True)
    ax.set_ylim(bottom=ymin[poly])

# bold shared y-label
fig.text(0.02, 0.5, "DSC (mW/mg)", va="center", rotation="vertical",
         fontsize=12, fontweight="bold")

# bold x-label
axes[-1].set_xlabel("Temperature (°C)", fontsize=12, fontweight="bold")

plt.tight_layout(rect=(0.05, 0, 1, 1))
plt.savefig("DSC_panels_highres.png", dpi=600, bbox_inches="tight")
plt.show()
```
